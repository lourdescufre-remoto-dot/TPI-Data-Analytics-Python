```python
#=========================
# Limpieza de datos ventas
#=========================
print("\nLimpieza de df_ventas")
print("-" * 50)

# Crear copia limpia
df_ventas_limpio = df_ventas.copy()

# Eliminamos duplicados
duplicados_antes = df_ventas_limpio.duplicated().sum()
print(f"Duplicados antes: {duplicados_antes}")
df_ventas_limpio.drop_duplicates(inplace=True)
print(f"Duplicados después: {df_ventas_limpio.duplicated().sum()}")

# Limpiar precio y convertir a float
df_ventas_limpio['precio'] = (
    df_ventas_limpio['precio']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

# Eliminar filas con valores nulos en precio o cantidad
df_ventas_limpio = df_ventas_limpio.dropna(subset=['precio','cantidad'])

# Convertir fechas a datetime
df_ventas_limpio['fecha_venta'] = pd.to_datetime(
    df_ventas_limpio['fecha_venta'],
    dayfirst=True,
    errors='coerce'
)

# Crear columna de importe
df_ventas_limpio['importe'] = df_ventas_limpio['precio'] * df_ventas_limpio['cantidad']

# Mostrar resultados
print("\nDatos limpios (primeras filas):")
display(df_ventas_limpio.head())

#=========================
# Limpieza de datos marketing
#=========================
print("\nLimpieza de df_marketing")
print("-" * 50)

# Crear copia limpia
df_marketing_limpio = df_marketing.copy()

# Revisar duplicados
duplicados_antes = df_marketing_limpio.duplicated().sum()
print(f"Duplicados antes: {duplicados_antes}")
df_marketing_limpio.drop_duplicates(inplace=True)
print(f"Duplicados después: {df_marketing_limpio.duplicated().sum()}")

# Convertir fechas a datetime
df_marketing_limpio["fecha_inicio"] = pd.to_datetime(
    df_marketing_limpio["fecha_inicio"], dayfirst=True, errors='coerce'
)
df_marketing_limpio["fecha_fin"] = pd.to_datetime(
    df_marketing_limpio["fecha_fin"], dayfirst=True, errors='coerce'
)

# Limpiar columna costo y convertir a float
df_marketing_limpio['costo'] = (
    df_marketing_limpio['costo']
    .astype(str)
    .str.replace('$', '', regex=False)
    .str.replace(',', '.', regex=False)
    .astype(float)
)

# Eliminar filas con valores nulos en columnas críticas
df_marketing_limpio = df_marketing_limpio.dropna(subset=['costo','fecha_inicio', 'fecha_fin'])

# Mostrar resultados
print("\nDatos limpios (primeras filas):")
display(df_marketing_limpio.head())

#=========================
# Limpieza de datos clientes
#=========================
print("\nLimpieza de df_clientes")
print("-" * 50)

# Crear copia limpia
df_clientes_limpio = df_clientes.copy()

# Revisar duplicados
duplicados_antes = df_clientes_limpio.duplicated().sum()
print(f"Duplicados antes: {duplicados_antes}")

# Revisar valores nulos
nulos_por_columna = df_clientes_limpio.isnull().sum()
print("Valores nulos por columna:")
print(nulos_por_columna)

# Asegurar tipos correctos
df_clientes_limpio["id_cliente"] = df_clientes_limpio["id_cliente"].astype(int)
df_clientes_limpio["edad"] = df_clientes_limpio["edad"].astype(int)
df_clientes_limpio["ingresos"] = df_clientes_limpio["ingresos"].astype(float)

# Mostrar primeras filas
print("\nPrimeras filas del dataset limpio:")
display(df_clientes_limpio.head())
```

```python
print ("=" * 50)
print ("\nAnálisis de ventas mensuales: Enero a Diciembre")
print ("=" * 50)

# Aseguramos que la fecha está en formato datetime
df_ventas_limpio['fecha_venta'] = pd.to_datetime(df_ventas_limpio['fecha_venta'], errors='coerce')

# Creo un diccionario de meses
meses_dict = {
    1:"Enero", 2:"Febrero", 3:"Marzo", 4:"Abril", 5:"Mayo", 6:"Junio", 7:"Julio", 8:"Agosto",
    9:"Septiembre", 10:"Octubre", 11:"Noviembre", 12:"Diciembre"
}

# Iniciar un diccionario para almacenar ventas mensuales
ventas_por_mes = {mes:0.0 for mes in range(1, 13)} # Se agrega .0 para asegurar flotantes

# Calcular ventas por mes
for index, row in df_ventas_limpio.iterrows():
    # Obtener el mes de la fecha de venta
    try: # Se añade un try/except para manejar NaT (fechas inválidas)
        mes_num = row["fecha_venta"].month
    except AttributeError:
        # Saltar si la fecha es inválida (NaT)
        continue

    # Sumar el monto del mes correspondiente
    ventas_por_mes[mes_num] += row['importe']


# Encontrar mes con más y menos ventas
# Usando max y min con una función key sobre los items del diccionario
mes_mayor = max(ventas_por_mes.items(), key=lambda x: x[1])
mes_menor = min(ventas_por_mes.items(), key=lambda x: x[1])

# Imprimir ventas mensuales
print("\nVentas por mes")
for mes_num, ventas in ventas_por_mes.items():
    print (f"{meses_dict[mes_num]}: ${ventas:,.2f}")

# Calcular ventas totales y promedio
ventas_totales = sum(ventas_por_mes.values())
promedio_mensual = ventas_totales / 12

#Imprimir resumen
print ("-" * 50)
print ("\nResumen de ventas anuales, promedio mensual, mes con más ventas y menos ventas")
print ("-" * 50)
print (f"Ventas totales anuales: ${ventas_totales:,.2f}")
print(f"Promedio mensual: ${promedio_mensual:,.2f}")
print(f"Mes con más ventas: {meses_dict[mes_mayor[0]]} (${mes_mayor[1]:,.2f})")
print(f"Mes con menos ventas: {meses_dict[mes_menor[0]]} (${mes_menor[1]:,.2f})")

# Creo un DataFrame de resumen de ventas por mes
ANIO_VENTAS = 2024


# Creo una lista de diccionarios para el DataFrame
datos_resumen = []
for mes_num, total_ventas in ventas_por_mes.items():
    fila = {
        'Año': ANIO_VENTAS,
        'Mes_Numero': mes_num,
        'Nombre_Mes': meses_dict[mes_num],
        'Total_Ventas': total_ventas
    }
    datos_resumen.append(fila)
df_resumen_final = pd.DataFrame(datos_resumen)
print("=" * 50)
print("\nDataFrame Resumen Final:")
print("=" * 50)

display(df_resumen_final)
```

## Ventas

Se filtraron y transformaron los datos de `df_ventas_limpio` para identificar los productos con mejor desempeño:

- Se calcularon **unidades vendidas** e **ingresos totales** por producto.
- Se seleccionaron productos con **ventas mínimas** (`MIN_VENTAS`) y los **Top-N por ingresos** (`TOP_N`).

```python
# Transformación productos de alto rendimiento

# Parametros
MIN_VENTAS = 130 #Minimo de unidades vendidas para considerar "Alto rendimiento"
TOP_N = 10 #Productos por ingreso

# Validar que existen las columnas necesarias
requeridas = {'producto', 'cantidad', 'importe'}
faltan = requeridas - set (df_ventas_limpio.columns)
if faltan:
  raise ValueError(f"Faltan columnas requeridas: {faltan}")

 # Agrupar por producto
tabla_por_producto = (
    df_ventas_limpio.groupby("producto", as_index=False)
    .agg(
        ingresos_totales=("importe", "sum"),
        unidades_totales=("cantidad", "sum"),
        ventas=("producto", "size")
    )
)

# Ordenar por ingresos descendente
tabla_ordenada = tabla_por_producto.sort_values("ingresos_totales", ascending=False).reset_index(drop=True)

# Ventas > MIN_VENTAS
buen_rendimiento_min_ventas = tabla_ordenada.loc[tabla_ordenada["ventas"] >= MIN_VENTAS]

# TOP_N por ingresos
if len(tabla_ordenada) >= TOP_N:
  umbral = float(tabla_ordenada.iloc[TOP_N -1 ]["ingresos_totales"])
  top_por_ingresos = tabla_ordenada.loc[tabla_ordenada["ingresos_totales"] >= umbral]
else:
  top_por_ingresos = tabla_ordenada.copy()


 # Resultados

print ("Productos con ventas > MIN_VENTAS")
display (buen_rendimiento_min_ventas)

print ("\nTop", TOP_N, "productos por ingresos")
display (top_por_ingresos)

# Resumen
print("=" * 50)
print ("\nResumen")
print("=" * 50)
print("Total productos distintos:", len(tabla_por_producto))
print("Productos con ventas > MIN_VENTAS:", len(buen_rendimiento_min_ventas))
print(f"Top-{TOP_N} por ingresos:", len(top_por_ingresos))
```

## Ventas por Categoría

Se utilizó el dataset `df_ventas_limpio` para analizar las ventas agrupadas por **categoría de producto**.  

Se calcularon los siguientes indicadores para cada categoría:  
- **Ingresos totales**: suma de los importes de todas las ventas de la categoría.  
- **Unidades vendidas**: cantidad total de productos vendidos.  
- **Transacciones**: cantidad de ventas registradas.  
- **Participación %**: proporción de ingresos de la categoría respecto al total de todas las categorías.  
- **Ticket promedio**: ingreso promedio por transacción dentro de la categoría.  

**Resumen de resultados:**  
- Total de categorías distintas: 6  
- Categoría con mayores ingresos: Electrodomésticos  
- Categoría con menores ingresos: Decoración  


```python
 # ================================
# Agrupar por categoría y calcular resumen
# ================================

# Agrupar y resumir
tabla_por_categoria = (
    df_ventas_limpio.groupby("categoria", as_index=False)
    .agg(
        ingresos=("importe", "sum"),
        unidades=("cantidad", "sum"),
        transacciones=("id_venta", "count")
    )
    .sort_values("ingresos", ascending=False)
    .reset_index(drop=True)
)

# Participación de cada categoría sobre el total
total_ingresos = tabla_por_categoria["ingresos"].sum()
tabla_por_categoria["participacion_%"] = (tabla_por_categoria["ingresos"] / total_ingresos * 100).round(2)

# Ticket promedio por transacción
tabla_por_categoria["ticket_promedio"] = (tabla_por_categoria["ingresos"] / tabla_por_categoria["transacciones"]).round(2)

# ================================
# Mostrar resultados
# ================================
print("=" * 50)
print("Ventas por Categoría")
print("=" * 50)
display(tabla_por_categoria)

# Resumen
print("=" * 50)
print("Resumen por categoría")
print("=" * 50)
print(f"Total categorías distintas: {len(tabla_por_categoria)}")
print(f"Categoría con más ingresos: {tabla_por_categoria.iloc[0]['categoria']} (${tabla_por_categoria.iloc[0]['ingresos']:,.2f})")
print(f"Categoría con menos ingresos: {tabla_por_categoria.iloc[-1]['categoria']} (${tabla_por_categoria.iloc[-1]['ingresos']:,.2f})")
```

# 4- Integración de Datos: Combinar los sets de datos de ventas y marketing para obtener una visión más amplia de las tendencias.

### Se unieron los **DataFrames** de **Ventas** y **Marketing** mediante `pd.merge_asof()` para realizar una **unión por rango de fechas**. Esto asocia cada venta con la campaña de marketing que estaba activa en la fecha de la transacción.

# Estructura del Resultado
## df_ventas_integrado: (2,312 filas, 13 columnas)

`Ventas (_x)`: Columnas del registro de venta (ej: producto_x, fecha_venta).

`Marketing (_y)`: Columnas de la campaña activa (ej:canal, costo).

### Esté DataFrame nos servira para tener una visualización más amplia de las tendencias

```python
 # Voy a utilizar el método merge para combinar los sets de datos
# Primero me fijo que las columnas de fecha sean datetime
df_ventas_limpio['fecha_venta'] = pd.to_datetime(df_ventas_limpio['fecha_venta'])
df_marketing_limpio['fecha_inicio'] = pd.to_datetime(df_marketing_limpio['fecha_inicio'])

# Ordeno los DataFrames
df_ventas_sorted = df_ventas_limpio.sort_values('fecha_venta')
df_marketing_sorted = df_marketing_limpio.sort_values('fecha_inicio')

# Ahora voy a realizar la unión con merge_asof
# Una cada venta con la campaña que está más cerca
df_ventas_integrado = pd.merge_asof(
    df_ventas_sorted,
    df_marketing_sorted,
    left_on = 'fecha_venta', # clave de venta
    right_on =  'fecha_inicio', #clave de inicio de cxampaña
    direction = 'backward' # Busca la campaña más cercana
)

# Filtrar que la venta ocurrió dentro del rango de la campaña
df_ventas_integrado = df_ventas_integrado[
    df_ventas_integrado['fecha_venta'] <= df_ventas_integrado['fecha_fin']].reset_index(drop=True)

# Muestro los resultados
print("-" * 50)
print("Integración de datos: Ventas con Campañas de Marketing")
print("-" * 50)
print (f"Filas: {df_ventas_integrado.shape[0]}, Columnas {df_ventas_integrado.shape[1]}")
df_ventas_integrado.head()
```

  ### Se generó un reporte agregando los datos por producto (groupby('producto_x')) para obtener métricas clave de negocio.

```python
# Agregación y calculo de KPIs

# 1. Agrupar el DataFrame integrado por producto y sumar ingresos/costos
reporte_rentabilidad = df_ventas_integrado.groupby('producto_x', as_index=False).agg(
    ventas_total=('importe', 'sum'),
    unidades_vendidas=('cantidad', 'sum'),
    inversion_total=('costo', 'sum'),
    total_transacciones=('id_venta', 'size')
)

# 2. Calcular los KPIs
# 'ROAS': Ingreso Total / Inversión Total
reporte_rentabilidad['ROAS'] = (reporte_rentabilidad['ventas_total'] / reporte_rentabilidad['inversion_total']).round(2)

# Ticket Medio = Ingreso Total / Total de Transacciones
reporte_rentabilidad['ticket_medio'] = (reporte_rentabilidad['ventas_total'] / reporte_rentabilidad['total_transacciones']).round(2)

# 3. Mostrar el ranking ordenado por ROAS (para ver los productos más rentables)
reporte_final = reporte_rentabilidad.sort_values('ROAS', ascending=False)

print("-" * 50)
print("\nReporte de Atribución y Rentabilidad por Producto:")
print("-" * 50)
reporte_final.head()
```

## Matriz de Correlación.
#### Permite identificar la fuerza y dirección de la relación entre pares de variables. El ROAS está fuertemente correlacionado con el Ticket Medio (+0.96).

```python
df_correlacion = reporte_final[[
    'ventas_total',
    'unidades_vendidas',
    'inversion_total',
    'total_transacciones',
    'ROAS',
    'ticket_medio'
]]
df_correlacion = df_correlacion.replace([np.inf, -np.inf], np.nan)
df_correlacion = df_correlacion.dropna()
matriz_corr = df_correlacion.corr()
plt.figure(figsize=(9, 6))
sns.heatmap(
    matriz_corr,
    annot=True,
    cmap='coolwarm',
    fmt=".2f",
    linewidths=.5,
    cbar_kws={'label': "Coeficiente de correlación"}
)
plt.title("Matriz de correlación de KPIs por producto", fontsize=16)
plt.show()
display(matriz_corr)
```
